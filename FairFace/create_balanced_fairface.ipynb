```python
save_folder = "data/"
```

```python
c = 0
for img_name in img_list:
    #print(img_name)
    save_name = save_folder + img_name.split("/")[-1]
    img = cv2.imread(img_name)
    cv2.imwrite(save_name, img)
    
    c += 1
    #break

print(c)
```
